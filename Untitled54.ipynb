```python
def manual_power(base: float, exponent: int) -> float:
    if exponent == 0:
        return 1.0
    if exponent < 0:
        return 1.0 / manual_power(base, -exponent)
    return base * manual_power(base, exponent - 1)

# Test it
print(manual_power(2, 3))
print(manual_power(5, -2))
```

```python
class ManualStack:
    def __init__(self):
        self.stack = []
        self.top_index = -1

    def push(self, item):
        self.stack += [item]
        self.top_index += 1

    def pop(self):
        if self.top_index == -1:
            return None
        item = self.stack[self.top_index]
        self.stack = self.stack[:self.top_index]
        self.top_index -= 1
        return item

def reverse_string(text: str) -> str:
    stack = ManualStack()
    for char in text:
        stack.push(char)
    reversed_text = ""
    while stack.top_index >= 0:
        reversed_text += stack.pop()
    return reversed_text

# Test it
print(reverse_string("engineer"))
```
